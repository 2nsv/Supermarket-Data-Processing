### Corrected data types & Checked duplicate records

```python
df['Sales'] = df['Sales'].str.replace('EGP',' ')
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Time'] = pd.to_timedelta(df['Time'])
```

### Sorted the data

```python
df = df.sort_values(
    ['Sales' , 'gross_income' , 'Rating' , 'gross_margin_percentage' , 'Unit_price' ] , ascending = False
)
```

```python
df.head(6)
```

### Created the `satisfied` column based on the rating

```python
df['Satisfaction_Rule'] = df['Rating'].apply(lambda Rating:'Satisfied' if Rating >= 7 else 'Not Satisfied')
df
```

### Create 3 columns ` (Day, Month, Year ) `

```python
df['Day'] = df['Date'].dt.day
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
```

```python
df['Day'] = pd.to_datetime(df['Day'])
df['Month'] = pd.to_datetime(df['Month'])
df['Year'] = pd.to_datetime(df['Year'])
```

#### 1. Total Revenue

```python
TotalRevenue = int (df['Sales'].sum())
print('Total Revenue : ' , TotalRevenue ,'EGP' )
```

```python
import matplotlib.pyplot as plt

plt.bar(['Total Revenue'], [TotalRevenue])

plt.title('Total Revenue')
plt.ylabel('Revenue (EGP)')

plt.text(0, TotalRevenue, f'{TotalRevenue:,} EGP',
         ha='center', va='bottom')

plt.show()
```

#### 2. Revenue by City

```python
print(df.groupby('City')['Sales'].sum().sort_values(ascending=False))
print('---> Highest Revenue City : Naypyitaw ' )
```

```python
import matplotlib.pyplot as plt

revenue_by_city = df.groupby('City')['Sales'].sum().sort_values(ascending=False)

plt.bar(revenue_by_city.index, revenue_by_city.values)

plt.title('Revenue by City')
plt.xlabel('City')
plt.ylabel('Revenue (EGP)')

for i, value in enumerate(revenue_by_city.values):
    plt.text(i, value, f'{value:,.0f}', ha='center', va='bottom')

plt.show()
```

#### 3. Profit by Branch

```python
print(df.groupby('Branch')['Sales'].sum().sort_values(ascending=False))
print('---> Most Profitable Branch : Giza ' )
```

```python
import matplotlib.pyplot as plt

sales_by_branch = df.groupby('Branch')['Sales'].sum().sort_values(ascending=False)

plt.bar(sales_by_branch.index, sales_by_branch.values)

plt.title('Sales by Branch')
plt.xlabel('Branch')
plt.ylabel('Sales (EGP)')

plt.show()
```

#### 4. Revenue and Profit by Product Category

```python
print(df.groupby('Product_line')['Sales'].sum().sort_values(ascending=False))
print('---> Top-Performing Category : Food and beverages ' )
```

```python
import matplotlib.pyplot as plt

sales_by_product = df.groupby('Product_line')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

bars = plt.bar(sales_by_product.index, sales_by_product.values)

plt.title('Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Sales (EGP)')

plt.xticks(rotation=45)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{bar.get_height():,.0f}',
        ha='center',
        va='bottom'
    )

plt.show()
```
